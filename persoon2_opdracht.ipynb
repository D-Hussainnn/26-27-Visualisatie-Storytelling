In dit notebook maak ik grafieken D, E en F. De opdrachten gaan over bosgebied in 2020 en temperatuurgegevens van de wereld en Nederland.


## Benodigde bibliotheken en dataset


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("forest_area_cleaned.csv")
df.head()
```

## Opdracht D – Bosgebied wereldwijd en per inkomensniveau in 2020

Ik vergelijk het wereldwijde bosgebied met de vijf inkomensgroepen in 2020.


```python
gebieden = [
    "World",
    "Low income",
    "Lower middle income",
    "Middle income",
    "Upper middle income",
    "High income"
]

bos_2020 = df[(df["year"] == 2020) & df["area_name"].isin(gebieden)].copy()
bos_2020["area_name"] = pd.Categorical(
    bos_2020["area_name"], categories=gebieden, ordered=True
)
bos_2020 = bos_2020.sort_values("area_name")

plt.figure(figsize=(10, 5))
plt.barh(bos_2020["area_name"].astype(str), bos_2020["forest_area"])
plt.title("Bosgebied wereldwijd en per inkomensniveau (2020)")
plt.xlabel("Bosgebied (km²)")
plt.ylabel("Gebied")
plt.ticklabel_format(style="plain", axis="x")
plt.tight_layout()
plt.show()
```

**Grafiekkeuze:** Ik heb een horizontaal staafdiagram gekozen, omdat je hiermee de bosgebieden in 2020 goed met elkaar kunt vergelijken.

**Clutter:** Ik heb alleen de zes gebieden, de titel en de namen van de assen laten zien. Zo blijft de grafiek overzichtelijk.

**Preattentive attributes:** De lengte van de balken maakt de verschillen in bosgebied snel zichtbaar.


## Opdracht E – Gemiddelde wereldtemperatuur (1995–2019)

Voor deze grafiek gebruik ik de jaarlijkse wereldwijde temperatuurafwijking van NASA. Dit is het verschil met het gemiddelde van 1951–1980, in graden Celsius. Het gaat dus om een temperatuurafwijking en niet om de absolute temperatuur.


```python
from temperature_data import prepare_temperature_data

global_temp, nederland_temp = prepare_temperature_data()
global_temp = global_temp[
    (global_temp["year"] >= 1995) & (global_temp["year"] <= 2019)
]

plt.figure(figsize=(10, 5))
plt.plot(global_temp["year"], global_temp["temperature_anomaly_c"], marker="o")
plt.title("Wereldwijde temperatuurafwijking (1995–2019)")
plt.xlabel("Jaar")
plt.ylabel("Temperatuurafwijking (°C)")
plt.xticks(range(1995, 2020, 2), rotation=45)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

**Grafiekkeuze:** Ik heb een lijngrafiek gekozen, omdat je hiermee de ontwikkeling van de temperatuur door de jaren heen kunt volgen.

**Clutter:** Ik heb alleen de jaren, temperatuurafwijking en een lichte horizontale rasterlijn gebruikt. Daardoor blijft de lijn goed zichtbaar.

**Preattentive attributes:** De lijn en de punten trekken de aandacht naar de ontwikkeling per jaar.

**Bron:** NASA GISTEMP. De waarden zijn temperatuurafwijkingen ten opzichte van 1951–1980.


## Opdracht F – Gemiddelde temperatuur per maand in Nederland

Voor deze boxplot gebruik ik dagelijkse gemiddelde temperaturen van KNMI-station De Bilt van 1995 tot en met 2019. Het gaat om metingen van één station en niet om het gemiddelde van alle plaatsen in Nederland.


```python
maanden = [
    "Januari", "Februari", "Maart", "April", "Mei", "Juni",
    "Juli", "Augustus", "September", "Oktober", "November", "December"
]

nl_temp = nederland_temp[
    (nederland_temp["year"] >= 1995) & (nederland_temp["year"] <= 2019)
].copy()
nl_temp["month_name"] = nl_temp["month"].map(dict(enumerate(maanden, start=1)))

plt.figure(figsize=(12, 6))
sns.boxplot(data=nl_temp, x="month_name", y="avg_temperature_c", order=maanden)
plt.title("Gemiddelde dagtemperatuur per maand in De Bilt (1995–2019)")
plt.xlabel("Maand")
plt.ylabel("Gemiddelde temperatuur (°C)")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

**Grafiekkeuze:** Ik heb een boxplot gekozen, omdat je hiermee de spreiding van de dagelijkse temperaturen per maand kunt vergelijken. Je ziet onder andere de mediaan en de spreiding.

**Clutter:** Ik heb de grafiek beperkt tot twaalf boxplots, één voor elke maand.

**Preattentive attributes:** De twaalf boxplots staan naast elkaar. Hierdoor vallen verschillen tussen koude en warme maanden snel op.

**Bron en beperking:** De temperatuurmetingen komen van KNMI-station De Bilt. De grafiek beschrijft dus De Bilt als meetlocatie en niet iedere plek in Nederland.
